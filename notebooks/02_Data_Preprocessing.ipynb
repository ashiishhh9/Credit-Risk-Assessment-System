```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler

import os
import warnings

warnings.filterwarnings("ignore")

# Display Settings
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/loan_default.csv")

df.head()
```

# 📄 Create Dataset Copy

Creating a copy of the original dataset ensures that the raw data remains unchanged throughout the preprocessing pipeline.

```python
data = df.copy()

data.head()
```

# 🗑 Remove Unnecessary Columns

The `LoanID` column is a unique identifier and does not contribute to predicting loan default.

Therefore, it is removed before model development.

```python
data.drop(columns=["LoanID"], inplace=True)

data.head()
```

# 🎯 Separate Features and Target

Machine learning models require the dataset to be separated into:

- Feature Matrix (X)
- Target Variable (y)

```python
X = data.drop(columns=["Default"])   # Take all columns except the target
y = data["Default"]   # Target column

print("Feature Matrix Shape :", X.shape)
print("Target Shape :", y.shape)
```

### Identify Numerical and Categorical Columns

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Features\n")
print(numerical_features)

print()

print("Categorical Features\n")
print(categorical_features)
```

# 🔠 One-Hot Encoding

Machine learning algorithms require numerical input.

Categorical variables are transformed using **One-Hot Encoding**, which creates binary columns for each category while avoiding any artificial ordering between categories.

```python
X = pd.get_dummies(
    X,
    columns=categorical_features,
    drop_first=True
)

X.head()
```

# 📏 Feature Scaling

Numerical features are standardized using **StandardScaler**.

Standardization improves the performance of algorithms such as Logistic Regression by ensuring all numerical features are on a comparable scale.

```python
scaler = StandardScaler()

X[numerical_features] = scaler.fit_transform(
    X[numerical_features]
)

X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features :", X_train.shape)

print("Testing Features :", X_test.shape)

print("Training Target :", y_train.shape)

print("Testing Target :", y_test.shape)
```

# 💾 Save Processed Dataset

The processed datasets are saved for reuse during model training and evaluation.

```python
# Create processed data directory

os.makedirs("../data/processed", exist_ok=True)

# Save processed files

X_train.to_csv("../data/processed/X_train.csv", index=False)

X_test.to_csv("../data/processed/X_test.csv", index=False)

y_train.to_csv("../data/processed/y_train.csv", index=False)

y_test.to_csv("../data/processed/y_test.csv", index=False)

processed_data = pd.concat([X, y], axis=1)

processed_data.to_csv(
    "../data/processed/cleaned_data.csv",
    index=False
)

print("Processed dataset saved successfully.")
```
